### Creating class blueprint

- Blueprint:-

  Blueprint is drawing of objects to creat.

```python
class Students():
    def __init__ (self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def walk(self):
        print("He is walking.")
    def talk(self):
        print("He is talking.")

student1 = Students("Rahat", "32y", "Male")
student1.name
print(student1.name)
student1.walk()
```
